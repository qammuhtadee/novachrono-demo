## Game progression

1. Would contain at maximum two-three sprites at a time at best, only one of the sprites will do movement at any given moment to do a certain task, and the player doesn't get to choose which one, to avoid complexity.
2. 3rd person view for general cases and to overlook the whole environment, 1st person view to only show the specific machinery being used at any moment and if something has to be done with it, like making sure of some measures are within the tolerable percentage.
3. There would be no specific levels but rather in-game Martian days passed as checkpoints, every single day it passes challenges would be harder, new issues would arise and you have to manage the four parameters (oxygen, water, power and radiation) at the same time and also keep a lookout for tradeoffs.
4. Controls would be the left and right arrow keys to walk, space button to access a machinery, and Esc or a physical back button within the game to go back, some other buttons to perform very specific functionality.
5. Sprites won't be a problem, yes we will make wacky sprites, they would look ugly but they would deliver the basic idea and also let people know that yes this is the proper identified machine. If we find art expertise, the map takes the textured look of Polus from Among Us.
6. There would be no scoreboard system. The game shall go on for a week (7 in game days) within these 7 days we would have to bring all four parameters to a sustainable threshold and the tradeoffs and stuff we will do would cancel each other and optimise the maximum possibility. There would be a performance marker as to how much well it was optimized as every player would have a different approach. Once the 7 days finish we will finish the game and give them a rating as to what could be improved to make it better, along with scientific tips, and say if they continue this every week, they will survive until the rescue mission comes and picks them back up, given proper energy and resources budget.
7. If they couldn't survive one of those 7 days, destroys a machine by failing to protect it, or any astronaut goes below 30% HP, then the in-game days would be checkpoints. The game would resume along with some tips as to how to solve this (we will not give direct hints but rather present them with scientific facts relevant to the particular challenge they are failing to solve as much as such that it becomes obvious to them what to do) on the day they failed and start from there with every baseline being set at where that day began. The same kind of fact, written as a thinking-bubble hint, also comes before each choice, as in the script.

## How we build the script

We have to first make a script, do some flowcharts which would have multiple branches of success methods, we will add at best three cases of success, but one of them would be the most optimised. Every other decision branch would lead to a game over or we won't allow certain combinations within the game, so that people can't go rogue and our game doesn't know what to do with this probably better or equal optimisation but just because it wasn't coded in our system, it won't be flagged as a win, and that would be a huge embarrassment for us. So we have to control the flow.

In short, before the Hackathon starts we need three things:

1. The script, with the challenges of each day and what raises or lowers the performance marker.
2. The flowcharts, with at best three success cases and one most optimised.
3. The information pond of verified facts from NASA data archives and 3rd party sources.</small>

# Base Introductions of game plot

<small>
Consider the game like a piece of short film. Two Astronauts have an unintended crash landing on Mars after some serious accidents. The team consists of the two member, an engineer and a scientist-medic. Right after landing, they take immediate shelter behind whatever is left of the now broken spacecraft, and after painstakingly fixing the radio and comms module by the engineer, they figure out the communication delay is excruciatingly long at around 15-22 minutes. Because the delay is this long, there is no point waiting for help from base, so the player must figure stuff out himself/herself.

And to their horrific surprise, Mars doesn't have a global magnetic field and only has a very thin atmosphere, making them very vulnerable to solar storm and cosmic radiation. If by chance the comms module remain operational during these calamities, the electronics inside can still get damaged unless they are put in safe mode.

On day 0, they figure out that they have some nanomaterials (inspiration taken from Iron Man) which they can use to build out reliable apparatus for any purposes, but only within a strict budget, so they can never build every apparatus on the same day. The permanent base to live on and take shelter is already standing when the game starts, made regardless of how much nanomaterials they had before the start, so the budget only counts for what they build from there on, and they start their new journey as a two buddy astronomical Robinson Crusoe duo.

Over time on day 1, they receive new messages from base, you have challenges to face which shows a common pattern every week, and if you can reinforce your learning to the point you can survive optimally managing resources for just a week, they can guarantee you can survive every week until rescue mission comes and picks you back up.

The 4 biggest concern you have to face is how to optimally maintain oxygen, water, radiation shielding of yourself, and managing power through nuclear energy and solar panels. For every research you do, you must go outside and make yourself vulnerable to cosmic radiation but you have to face this tradeoff and risk your health meter for a necessary invention needed to upgrade your facilities. Mining ice goes the same way, and right after a dust storm the power supply equipments must be regulated and cleaned from the outside, so every one of these trips costs health too. On days that you have a solar storm, it's a no-no to outside at all, and every non-essential device that uses electrical power goes into safe mode while only life support keeps running.

To create oxygen, we can split water which comes from the underground ice we mine. Martian dust, on the other hand, can cover the solar panels, which makes power a real challenge, though the dust itself isn't a parameter concern.

The scientist takes the duty of life support (basically water + oxygen), the engineer takes duty of optimizing radiation shielding, and proper budgeting of power, and also create newer areas where power would be used for the betterment of the now permanent outpost.

The four meters are the real concern, and the health of each astronaut simply follows them: oxygen depleting and radiation exposure increasing both proportionally lower the astronaut's HP, and no astronaut shall go below 30% HP, if they want to survive, otherwise they get weak and battered down, facilitating them useless to the challenge and basically making the game over.

Each day is a challenge and each passed day is a check point. Before making any choices, we will be given a hint like a pop up or thinking bubble as if the astronaut learned this in his academy training to make sure the player playing this game is inspired to make the proper choice, and if they make any wrong choice, a hint will be shown on the game over or after penalty receiving screen in order to make sure they understand where they went wrong and do not make the same mistakes again. For every hint we say we will use, referenced and dated data from DONKI, OSDR, Perseverance and Artemis provided as per the archived NASA data.

The four parameters we have to optimize as well as make sure every astronaut has proper HP will have to be maintained and within the 7 days, it would go to a sustainable level. Work will be divided between the two astronauts, and to include diversity, we will make the scientist a female astronaut.

The game is played in 3rd person, where the left and right buttons make the astronaut walk, and on certain days only one character can be controlled at a time, with the player not allowed to choose which one to navigate, in order to avoid complexity. The outpost map takes its textured look from the Polus map of Among Us, and we will need art expertise for it.

Each of the astronaut will have a log book to automatically log what challenges they faced each day and what was their solution, and the base suggestion of the optimum solution and they can access the log book anyday. At the end of the game after it gets finished, the player would be encouraged to check their finalized log book and see their work come out beautifully as a learning outcome in the journey of 7 in game days.

The parameters can never go down a certain threshold, and the hints may look like this, it could be like,

> *"Hint: in 2021, Perseverance's MOXIE split the carbon dioxide in the Martian air and made oxygen in the following manner, should we go out now?"*</small>
